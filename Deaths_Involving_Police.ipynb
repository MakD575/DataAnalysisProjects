## Data Cleaning

### Checking for Missing Values and Duplicates

```python
df_hh_income.fillna({'Median Income': 0}, inplace=True)
df_hh_income.info()
```

```python
df_fatalities.dropna(inplace=True)
df_fatalities.info()
```

### Fixing Data Types

```python
df_pct_poverty['poverty_rate'] = df_pct_poverty['poverty_rate'].apply(lambda x:str(x).replace('-','0')).astype('float64')
df_pct_poverty.info()
```

```python
df_pct_completed_hs['percent_completed_hs'] = df_pct_completed_hs['percent_completed_hs'].apply(lambda x:str(x).replace('-','0')).astype('float64')
df_pct_completed_hs.info()
```

```python
df_share_race_by_city['share_white'] = df_share_race_by_city['share_white'].apply(lambda x:str(x).replace('(X)', '0')).astype('float64')
df_share_race_by_city['share_black'] = df_share_race_by_city['share_black'].apply(lambda x:str(x).replace('(X)', '0')).astype('float64')
df_share_race_by_city['share_native_american'] = df_share_race_by_city['share_native_american'].apply(lambda x:str(x).replace('(X)', '0')).astype('float64')
df_share_race_by_city['share_asian'] = df_share_race_by_city['share_asian'].apply(lambda x:str(x).replace('(X)', '0')).astype('float64')
df_share_race_by_city['share_hispanic'] = df_share_race_by_city['share_hispanic'].apply(lambda x:str(x).replace('(X)', '0')).astype('float64')
df_share_race_by_city.info()
```

```python
df_fatalities['date'] = pd.to_datetime(df_fatalities['date'])
df_fatalities.info()
```

## Visualizing Poverty Rate by State

```python
pct_poverty_by_state = df_pct_poverty.groupby('Geographic Area')['poverty_rate'].mean().sort_values(ascending=False)
pct_poverty_by_state.head()
```

```python
pct_poverty_by_state.plot(
    kind='bar',
    figsize=(15, 10),
    rot=45,
    color='#fc9272'
)

plt.title('Poverty Rate by State', fontsize=20)
plt.ylabel('Poverty Rate (%)', fontsize=15)
plt.xlabel('State', fontsize=15)
plt.yticks(np.arange(0, 32, 2))
plt.grid(axis='y')
plt.show()
```

From the chart above we can see that the State of Mississippi has the highest poverty rate at 26.8% while the state of New Jersey has the lowest at 8.16%.

## Visualizing High School Graduation Rate by State

```python
pct_grads_by_state = df_pct_completed_hs.groupby('Geographic Area')['percent_completed_hs'].mean().sort_values()
pct_grads_by_state.head()
```

```python
pct_grads_by_state.plot(
    kind='bar',
    figsize=(15, 10),
    rot=45,
    color='#31a354'
)

plt.yticks(np.arange(70, 105, 5))
plt.title('High School Graduation Rate by State', fontsize=20)
plt.ylabel('High School Graduation Rate (%)', fontsize=15)
plt.xlabel('State', fontsize=15)
plt.grid(axis='y')
plt.show()
```

From the chart above we can see that the State of Texas has the lowest average graduation rate at 74.09% while the State of Massachusetts is highest at 92.03%.

## Visualizing the Relationship Between Poverty Rates and High School Graduation Rates

```python
pct_grads_by_state.sort_index(inplace=True)
pct_poverty_by_state.sort_index(inplace=True)
pct_grads_by_state.index == pct_poverty_by_state.index
```

```python
plt.figure(figsize=(20,10), dpi=200)
plt.title('Relationship between Poverty and High School Graduation Rates', fontsize=20)

poverty_axis = plt.gca()
grads_axis = poverty_axis.twinx()

poverty_axis.set_xlabel('State', fontsize=15)
poverty_axis.set_ylabel('Poverty Rates (%)', color='#fc9272', fontsize=15)
grads_axis.set_ylabel('High School Graduation Rates', color='#31a354', fontsize=15)

poverty = poverty_axis.plot(pct_poverty_by_state.index, pct_poverty_by_state, color='#fc9272', linestyle='dashdot', marker='s')
grads = grads_axis.plot(pct_grads_by_state.index, pct_grads_by_state, color='#31a354', linestyle='dashed', marker='o')

plt.grid()
plt.show()
```

### Visualizing Poverty and High School Graduation Rates using Kernel Density Estimate(KDE)

```python
poverty_graduation_comparison = pd.concat([pct_grads_by_state, pct_poverty_by_state], axis=1)
poverty_graduation_comparison
```

```python
sns.jointplot(
    data = poverty_graduation_comparison,
    y='percent_completed_hs',
    x='poverty_rate',
    kind='kde',
    color='#056875'
)

plt.xlabel('Poverty Rate', fontsize=15)
plt.ylabel('Percentage of Grads', fontsize=15)

plt.show()
```

### Visualizing Poverty and High School Graduation Rates Using Linear Regression

```python
sns.regplot(
    data=poverty_graduation_comparison,
    y='percent_completed_hs',
    x='poverty_rate',
    color='#056875'
)

plt.xlabel('Poverty Rate', fontsize=15)
plt.ylabel('Percentage of Grads', fontsize=15)

plt.show()
```

As we can see there is a clear correlation between the poverty rate and the graduation rate of any given state. As poverty rates go down the graduation rates go up and vice versa.

## Visualization of the Racial Makeup of Each US State

```python
race_by_state = df_share_race_by_city.groupby('Geographic area')[['share_white', 'share_black', 'share_asian', 'share_hispanic', 'share_native_american']].mean()
race_by_state
```

```python
sum_per_state = race_by_state['share_asian'] + race_by_state['share_black'] + race_by_state['share_hispanic'] + race_by_state['share_native_american'] + race_by_state['share_white']  
race_by_state['share_asian'] = race_by_state['share_asian'] / sum_per_state * 100
race_by_state['share_black'] = race_by_state['share_black'] / sum_per_state * 100
race_by_state['share_hispanic'] = race_by_state['share_hispanic'] / sum_per_state * 100
race_by_state['share_native_american'] = race_by_state['share_native_american'] / sum_per_state * 100
race_by_state['share_white'] = race_by_state['share_white'] / sum_per_state * 100
```

```python
race_by_state.plot(
    kind='bar',
    figsize = (20, 15),
    rot=0,
    stacked=True,
    ylim=(0,100),
)

plt.legend(['White', 'Black', 'Asian', 'Hispanic', 'Native American'], loc='lower left')
plt.title('Racial Makeup By State', fontsize=20)
plt.xlabel('State', fontsize=15)
plt.ylabel('Percentage by Race', fontsize=15)
```
